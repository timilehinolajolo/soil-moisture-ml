```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# Phase 0: Data Generation\n",
    "Creating a synthetic dataset that mimics the environmental conditions and gravimetric soil moisture readings at FUNAAB."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "import numpy as np\n",
    "import pandas as pd\n",
    "import os\n",
    "from datetime import datetime, timedelta\n",
    "\n",
    "np.random.seed(42)\n",
    "num_samples = 300\n",
    "\n",
    "start_date = datetime(2026, 9, 1)\n",
    "dates = [start_date + timedelta(days=i//10) for i in range(num_samples)]\n",
    "times = [f\"{8 + (i%10)}:00\" for i in range(num_samples)]\n",
    "\n",
    "temperature = np.random.uniform(24.0, 33.0, num_samples).round(1)\n",
    "humidity = np.random.uniform(70.0, 95.0, num_samples).round(1)\n",
    "rainfall = np.random.exponential(scale=3.0, size=num_samples)\n",
    "rainfall = np.where(rainfall < 1.5, 0, rainfall).round(1) \n",
    "depth = np.random.choice([10, 20, 30], num_samples)\n",
    "\n",
    "base_moisture = 18.0\n",
    "soil_moisture = (\n",
    "    base_moisture \n",
    "    + (rainfall * 0.8) \n",
    "    - ((temperature - 24) * 0.4) \n",
    "    + (humidity * 0.05) \n",
    "    + (depth * 0.15) \n",
    "    + np.random.normal(0, 1.5, num_samples)\n",
    ")\n",
    "\n",
    "soil_moisture = np.clip(soil_moisture, 8.5, 32.0).round(2)\n",
    "\n",
    "df = pd.DataFrame({\n",
    "    'Date': [d.strftime('%Y-%m-%d') for d in dates],\n",
    "    'Time': times,\n",
    "    'Temperature_C': temperature,\n",
    "    'Relative_Humidity_Pct': humidity,\n",
    "    'Rainfall_mm': rainfall,\n",
    "    'Sampling_Depth_cm': depth,\n",
    "    'Soil_Moisture_Pct': soil_moisture\n",
    "})\n",
    "\n",
    "os.makedirs('../data/synthetic', exist_ok=True)\n",
    "df.to_csv('../data/synthetic/funaab_synthetic_data.csv', index=False)\n",
    "print(\"Success! Synthetic dataset generated and saved.\")"
   ]
  }
 ],
 "metadata": {},
 "nbformat": 4,
 "nbformat_minor": 5
}
```
